## IMPORT LIBRARY

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Membaca Gambar

```python
img = cv2.imread('backlight.jpeg')
```

```python
[baris, kolom] = img.shape[:2]
```

## KONVERSI TO GRAYSCALE

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

## Show Citra Asli Dan Gray

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 5))
axs[0].imshow(rgb)
axs[0].set_title('Citra Asli (Backlight)')
axs[0].axis('off')

axs[1].imshow(gray, cmap='gray')
axs[1].set_title('Grayscale')
axs[1].axis('off')
plt.show()
```

## OPERASI PIKSEL GRAY PENCERAHAN

```python
beta = 50
citra_cerah = np.zeros((baris, kolom))

for x in range(baris):
    for y in range(kolom):
        gyx = gray[x, y] + beta
        if gyx > 255:          
            gyx = 255
        citra_cerah[x, y] = gyx
citra_cerah = citra_cerah.astype(np.uint8)
```

## OPERASI PIKSEL GRAY KONTRAS

```python
alpha = 1.8
citra_kontras = np.zeros((baris, kolom))

for x in range(baris):
    for y in range(kolom):
        gyx = gray[x, y] * alpha
        if gyx > 255:          
            gyx = 255
        citra_kontras[x, y] = gyx
citra_kontras = citra_kontras.astype(np.uint8)
```

## OPERASI PIKSEL BRIGHTNES & KONTRAS

```python
alpha2 = 1.4
beta2 = 35
citra_gabungan = np.zeros((baris, kolom))

for x in range(baris):
    for y in range(kolom):
        gyx = alpha2 * gray[x, y] + beta2
        if gyx > 255:         
            gyx = 255
        citra_gabungan[x, y] = gyx
citra_gabungan = citra_gabungan.astype(np.uint8)
```

## MENAMPILKAN SEMUA HASIL OPERASI

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Grayscale Asli
axs[0, 0].imshow(gray, cmap='gray')
axs[0, 0].set_title('Grayscale Asli')
axs[0, 0].axis('off')

# Grayscale Dipercerah
axs[0, 1].imshow(citra_cerah, cmap='gray')
axs[0, 1].set_title(f'Dipercerah (+50)')
axs[0, 1].axis('off')

# Grayscale Dikontras
axs[1, 0].imshow(citra_kontras, cmap='gray')
axs[1, 0].set_title(f'Dikontras (x1.8)')
axs[1, 0].axis('off')

# Grayscale Gabungan 
axs[1, 1].imshow(citra_gabungan, cmap='gray')
axs[1, 1].set_title(f'Gabungan')
axs[1, 1].axis('off')

plt.tight_layout()
plt.show()
```
